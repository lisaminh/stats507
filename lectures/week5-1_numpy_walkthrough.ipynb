## 1. Lists vs. arrays · slide 6

```python
values = [1, 2, 3]
a = np.array(values)
# Lists repeat; arrays multiply each numerical value.
print("List * 2: ", values * 2)
print("Array * 2:", a * 2)
# List addition concatenates; array addition adds matching entries.
print("List + list:  ", values + values)
print("Array + array:", a + a)
```

### A calculation on every value

```python
prices = [10, 20, 30]
# The comprehension visits one price at a time.
print([price * 1.08 for price in prices])
# Vectorization: apply the same 8% increase to the entire array.
print(np.array(prices) * 1.08)
```

## 5. Slicing · slide 18
**Predict:** Do `a[0]` and `a[0:1]` have the same shape?

```python
a = np.arange(1, 13).reshape(3, 4)
print(a)
# Rows 0 and 1; columns 1 and 2. Slice stops are excluded.
print("Slice:\n", a[:2, 1:3])
# The colon keeps all rows; -1 selects the final column.
print("Last column:", a[:, -1])
# An integer index removes an axis; a slice keeps that axis.
print("Shapes:", a[0].shape, a[0:1].shape)
```

### 1D slices: start, stop, step

```python
line = np.arange(10)
print("Middle:", line[2:7])
# Omitted start/stop span the full axis; step=2 skips alternate entries.
print("Every other:", line[::2])
# Negative indices count from the end.
print("Last three:", line[-3:])
# A negative step reverses traversal.
print("Reverse:", line[::-1])
```

### 2D slices and assignment

```python
grid = np.arange(1, 13).reshape(3, 4)
# First selector controls rows; second controls columns.
print("Every other column:\n", grid[:, ::2])
print("Reverse row order:\n", grid[::-1, :])
print("Column shapes:", grid[:, 1].shape, grid[:, 1:2].shape)
# Assign one value to a block: first two rows, last two columns.
grid[:2, -2:] = 0
print("After block assignment:\n", grid)
```

```python
a = np.arange(1, 13).reshape(3, 4)
# Basic slicing shares the original data rather than making a new copy.
view = a[:2, 1:3]
# copy() creates independent storage.
copied = a[:2, 1:3].copy()
# This location corresponds to a[0, 1], so the original also changes.
view[0, 0] = 99
print("Original changed:\n", a)
print("Independent copy:\n", copied)
```

## 7. Boolean operations · slides 21–22
Use `&`, `|`, and `~` for Boolean arrays; parenthesize comparisons.

```python
x = np.arange(10)
print("All > 7?", np.all(x > 7))
print("Any > 7?", np.any(x > 7))
# Stack two conditions: shape (2, 10).
conditions = np.array([x > 7, x < 2])
print("Conditions:\n", conditions)
# Reduce across the two conditions: one result for each value of x.
print("axis=0:", np.any(conditions, axis=0))
# Reduce across ten values: one result for each condition.
print("axis=1:", np.any(conditions, axis=1))
```

```python
print(x[np.logical_and(x > 3, x < 7)])
# Parentheses build each comparison before & combines the masks.
print(x[(x > 3) & (x < 7)])
print(x[(x < 3) | (x > 7)])
# XOR keeps entries satisfying exactly one of the conditions.
print(x[np.logical_xor(x > 3, x < 7)])
# ~ reverses each Boolean value; here it keeps x <= 3.
print(x[~(x > 3)])
```

### Combine conditions; keep or replace

```python
temperatures = np.array([-5, 12, 25, 40])
comfortable = (temperatures >= 10) & (temperatures <= 30)
print("Keep:", temperatures[comfortable])
# where(condition, if_true, if_false) keeps the original array shape.
print("Label all:", np.where(comfortable, "comfortable", "outside range"))
print("Clip negatives:", np.where(temperatures < 0, 0, temperatures))
```

### all / any: one decision per student

```python
scores = np.array([[80, 90], [60, 75], [95, 85]])
passed = scores >= 70
print(passed)
# axis=1 combines exam results within each student row.
print("Passed every exam:", passed.all(axis=1))
print("Passed at least one:", passed.any(axis=1))
# axis=0 combines students, leaving one decision per exam.
print("Everyone passed each exam:", passed.all(axis=0))
print("Students passing both:\n", scores[passed.all(axis=1)])
```

## 10. Vector and matrix operations · slides 27–29

```python
x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])
v = np.array([9, 10])
w = np.array([11, 12])
# For 1D vectors, dot multiplies matching entries and sums them.
print("Vector dot product:", np.dot(v, w))
# Each matrix row is dotted with the vector.
print("Matrix @ vector:", x @ v)
print("Elementwise:\n", x * y)
# For 2D arrays, @ combines rows of the left array with columns of the right.
print("Matrix product:\n", x @ y)
```

### Inner, outer, and cross products

```python
u = np.array([1, 2, 3])
v3 = np.array([4, 5, 6])
print("Inner:", np.inner(u, v3))
# Outer product entry (i, j) is u[i] * v3[j]; shape (3, 3).
print("Outer:\n", np.outer(u, v3))
# The 3D cross product is perpendicular to both input vectors.
print("Cross:", np.cross(u, v3))
```

### Non-square matrix multiplication

```python
A = np.arange(1, 7).reshape(2, 3)
B = np.arange(1, 13).reshape(3, 4)
print("Input shapes:", A.shape, B.shape)
# (2, 3) @ (3, 4) gives (2, 4); the inner dimensions must match.
print("A @ B:\n", A @ B)
print("Output shape:", (A @ B).shape)
print("First entry:", A[0] @ B[:, 0])
# B @ A fails because its inner dimensions are 4 and 2.
try:
    B @ A
except ValueError as error:
    print(error)
```

```python
# For a 2D array, .T swaps the row and column axes.
print("Transpose:\n", x.T)
# flatten() returns an independent 1D copy in row order.
print("Flatten:", x.flatten())
print("Determinant:", np.linalg.det(x))
# An inverse requires a square, nonsingular matrix.
print("Inverse:\n", np.linalg.inv(x))
```

### Solve a linear system

```python
A = np.array([[2., 1.], [1., 3.]])
b_rhs = np.array([5., 7.])
# Solve A @ solution = b_rhs directly, without computing an inverse.
solution = np.linalg.solve(A, b_rhs)
print("Solution:", solution)
print("A @ solution:", A @ solution)
# Check every reconstructed entry with a floating-point tolerance.
print("Correct?", np.allclose(A @ solution, b_rhs))
```

### A 1D transpose stays 1D

```python
vector = np.array([1, 2, 3])
print("vector:", vector.shape, "vector.T:", vector.T.shape)
# A 1D array has no separate row/column axes to swap; insert an axis first.
column_vector = vector[:, None]
print("Column:", column_vector.shape, "transposed:", column_vector.T.shape)
```

## 11. Broadcasting · slides 31–34
Compare shapes **from the right**: dimensions must match or one must be 1.

### Scalar, column offsets, row offsets

```python
table = np.arange(12).reshape(3, 4)
print("Original:\n", table)
# A scalar is reused at every position.
print("Add scalar:\n", table + 10)
# Shape (4,) aligns with the last axis of (3, 4).
print("Add per column:\n", table + np.array([10, 20, 30, 40]))
# Shape (3, 1) lets one offset repeat across each row.
print("Add per row:\n", table + np.array([100, 200, 300])[:, None])
```

```python
a = np.array([[0, 0, 0], [10, 10, 10], [20, 20, 20], [30, 30, 30]])
b = np.array([0, 1, 2])
print("Shapes:", a.shape, b.shape)
# Treat (3,) as (1, 3); reuse b across the four rows.
print(a + b)
```

```python
column = np.array([0, 10, 20, 30])[:, None]
row = np.array([0, 1, 2])[None, :]
print("Shapes:", column.shape, row.shape)
# (4, 1) + (1, 3) gives (4, 3): one sum for each row-column pair.
print(column + row)
```

```python
# Predict: will (4, 3) + (2, 3) work?
bad = np.array([[0, 1, 2], [3, 4, 5]])
# The last axes match, but 4 and 2 are incompatible; catch the demo error.
try:
    a + bad
except ValueError as error:
    print(error)
```

### Predict which shapes work

```python
shape_pairs = [((4, 3), (3,)), ((4, 3), (4,)),
               ((4, 3), (4, 1)), ((4, 1), (1, 3)),
               ((2, 3, 4), (3, 1))]
# Compare dimensions from the right; missing leading dimensions act like 1.
for left, right in shape_pairs:
    try:
        print(left, "+", right, "->", np.broadcast_shapes(left, right))
    except ValueError:
        print(left, "+", right, "-> incompatible")
```

### Center and standardize columns

```python
data = np.array([[10., 100.], [20., 140.], [30., 180.]])
# Reduce down rows to obtain one mean per column: shape (2,).
means = data.mean(axis=0)
# One standard deviation per column; NumPy uses ddof=0 by default.
stds = data.std(axis=0)
# Subtract the matching column mean from every observation.
centered = data - means
# Divide by each column's std; constant columns would need separate handling.
standardized = centered / stds  # Both columns have nonzero std
print("Column means:", means)
print("Centered:\n", centered)
print("Standardized:\n", standardized)
print("Check means / stds:", standardized.mean(axis=0), standardized.std(axis=0))
```

### A valid shape can still give the wrong meaning

```python
square = np.array([[1, 2], [10, 20]])
row_offsets = np.array([100, 200])
# Both operations are valid, but shape determines which axis gets the offsets.
print("1D offsets act on columns:\n", square + row_offsets)
print("Column-shaped offsets act on rows:\n", square + row_offsets[:, None])
```

## 12. Distances without loops · slide 35
One row = one point. Find distances between corresponding rows.

```python
# Each array contains five points; each row holds (x, y).
l1 = np.array([[0, 0], [2, 0], [4, 0], [6, 0], [8, 0]])
l2 = np.arange(10).reshape(5, 2)
print("Line 1:\n", l1)
print("Line 2:\n", l2)
```

```python
distances_loop = []
# zip pairs corresponding rows, rather than every possible pair.
for p, q in zip(l1, l2):
    distances_loop.append(np.sqrt((p[0] - q[0])**2 + (p[1] - q[1])**2))

# Subtract coordinates within each point pair: shape (5, 2).
diff = l1 - l2
# Square coordinate differences and sum across x/y: shape (5,).
squared_distances = (diff**2).sum(axis=1)
# Take a square root to convert squared distances to Euclidean distances.
distances = np.sqrt(squared_distances)
print("Shapes:", diff.shape, squared_distances.shape, distances.shape)
print("Loop:", distances_loop)
print("Vectorized:", distances)
print("Same?", np.allclose(distances_loop, distances))
```

### Extension: all pairwise distances

```python
points = np.array([[0., 0.], [3., 0.], [0., 4.]])
# (3, 1, 2) minus (1, 3, 2) broadcasts to (3, 3, 2).
delta = points[:, None, :] - points[None, :, :]
# Reduce the coordinate axis; entry (i, j) is the distance from point i to j.
pairwise = np.sqrt((delta**2).sum(axis=2))
print("Intermediate shape:", delta.shape)  # Point i x point j x coordinate
print("Pairwise distances:\n", pairwise)
```

```python
# A fixed seed makes this demo reproducible when the cell is rerun.
rng = np.random.default_rng(0)
dice = rng.integers(1, 7, size=10)          # 1 through 6
errors = rng.normal(loc=0, scale=1, size=10) # scale = standard deviation
print("Dice:", dice)
print("Normal samples:", errors)
print("Mean:", dice.mean())
# ddof=0 uses N in the variance denominator; ddof=1 uses N-1.
print("Std (ddof=0):", dice.std())
print("Std (ddof=1):", dice.std(ddof=1))
print("25th percentile:", np.percentile(dice, 25))
```
